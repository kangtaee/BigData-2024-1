```python
url = 'http://openapi.seoul.go.kr:8088/71705256767274303132324777687558/json/TvEmgcHospitalInfo/1/100/'

response = requests.get(url)
print(response.content)
```

```python
data = response.json()
data
```

```python
import pandas as pd

temp = []
for d in data['TvEmgcHospitalInfo']['row']:
  temp.append(d)
uni = pd.DataFrame(temp)
uni
```

```python
import folium
anMap = folium.Map(location=[37.88988548,127.1371472],
                    zoom_control=False,
                    zoom_start=11,
                    control_scale= True)
anMap
```

```python

from folium.plugins import MarkerCluster
mc = MarkerCluster().add_to(anMap)

for n, row in uni.iterrows():
  folium.Marker(location=[row['WGS84LAT'], row['WGS84LON']], popup = folium.Popup(row['DUTYEMCLSNAME'], max_width=200)).add_to(mc)

anMap
```
